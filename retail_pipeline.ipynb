```python
monthly_sales['3_Month_Volatility_%'] = (
    monthly_sales['MoM_Growth_%']
    .rolling(window=3)
    .std()
)

monthly_sales.head(10)
```

```python
import requests

url = "https://www150.statcan.gc.ca/n1/en/tbl/csv/20100056-eng.zip"

zip_path = r"D:\Retail Analytics with Statistics  Power BI Project\Data\20100056-eng.zip"

response = requests.get(url)

with open(zip_path, "wb") as file:
    file.write(response.content)

print("Download complete")
```

```python
import zipfile
import os

extract_folder = r"D:\Retail Analytics with Statistics  Power BI Project\Data\20100056-eng"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)

print("Extraction complete")
```

```python
import pandas as pd
import os

csv_path = os.path.join(
    extract_folder,
    "20100056-eng.csv"
)

df = pd.read_csv(csv_path)

print(df.shape)
df.head()
```

```python
df['REF_DATE'] = pd.to_datetime(df['REF_DATE'], format='%Y-%m')

df_clean = df.dropna(subset=['VALUE']).copy()

df_clean = df_clean[
    [
        'REF_DATE',
        'GEO',
        'North American Industry Classification System (NAICS)',
        'Sales',
        'Adjustments',
        'UOM',
        'SCALAR_FACTOR',
        'VALUE'
    ]
]

df_clean['Sales_Dollars'] = df_clean['VALUE'] * 1000

print(df_clean.shape)
```

```python
monthly_sales = (
    df_clean[
        (df_clean['GEO'] == 'Canada') &
        (df_clean['Sales'] == 'Total retail sales') &
        (df_clean['Adjustments'] == 'Seasonally adjusted') &
        (df_clean['North American Industry Classification System (NAICS)'] == 'Retail trade [44-45]')
    ]
    [['REF_DATE', 'Sales_Dollars']]
    .sort_values('REF_DATE')
    .copy()
)

monthly_sales['MoM_Growth_%'] = (
    monthly_sales['Sales_Dollars'].pct_change() * 100
)

monthly_sales['YoY_Growth_%'] = (
    monthly_sales['Sales_Dollars'].pct_change(periods=12) * 100
)

monthly_sales['3_Month_Moving_Avg'] = (
    monthly_sales['Sales_Dollars'].rolling(window=3).mean()
)

monthly_sales['12_Month_Moving_Avg'] = (
    monthly_sales['Sales_Dollars'].rolling(window=12).mean()
)

monthly_sales['3_Month_Volatility_%'] = (
    monthly_sales['MoM_Growth_%'].rolling(window=3).std()
)

monthly_sales['MoM_Z_Score'] = (
    monthly_sales['MoM_Growth_%'] - monthly_sales['MoM_Growth_%'].mean()
) / monthly_sales['MoM_Growth_%'].std()

monthly_sales['Anomaly_Flag'] = (
    monthly_sales['MoM_Z_Score'].abs() >= 2
)

monthly_sales.tail()
```

```python
output_path = r"D:\Retail Analytics with Statistics  Power BI Project\Data\monthly_sales_analysis.csv"

monthly_sales.to_csv(output_path, index=False)

print("Power BI file updated")
```
